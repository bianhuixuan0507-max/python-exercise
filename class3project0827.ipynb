```python
df['BMI_level'] = pd.cut(df["BMI"],
    bins = [0,24.5,30,50],
    labels = ['thin','normal','fat'])
```

```python
df.head()
```

```python
df['SkinThickness'] =  df['SkinThickness'].replace(0, pd.NA)
df['SkinThickness'] =  df['SkinThickness'].fillna(df['SkinThickness'].median())
# Skinthickness can not be a 0 value. It must be a positive number
```

```python
df['Insulin'] =  df['Insulin'].replace(0, pd.NA)
df['Insulin'] =  df['Insulin'].fillna(df['Insulin'].median())
# Insulin is a must-need hormone in human, it can not be 0.
```

```python
df['Age_group'] = pd.cut(df['Age'],
                         bins = [18,45,50,80],
                         labels= ['Young','Middle Age','Old']

)
```

```python
df.head()
```

```python
df.groupby('Outcome')['Glucose'].median()
#明显看出 确诊糖尿病的人群血糖中位数比未确诊的要高30
```

```python
df.groupby('Age_group')['Glucose'].agg(['mean','median'])
#   年龄越大，血糖的水平越高，青年和中年变化不大，老年血糖明显增大
```
