```python
class Patient:
    def __init__(self, name, glucose, BMI):
        self.name = name
        self.glucose = glucose
        self.BMI = BMI


    def high_glucose(self,normal= 70):
        if self.glucose > normal:
            return True


    def summary(self):
        print(f'Name is {self.name}')
        print(f'Glucose is {self.glucose}')
        print(f'BMI is {self.BMI}')
        self.high_glucose()

    def BMI_check(self):
        if self.BMI < 18.5:
            return 'thin'
        elif self.BMI < 24:
            return 'normal'
        else:
            return 'fat'

    def risk_note(self):
        """综合血糖和BMI给出风险提示"""
        is_high_glucose = self.high_glucose()
        bmi_status = self.BMI_check()

        # 根据不同组合给出提示
        if is_high_glucose and bmi_status == "fat":
            return "high risk"
        elif is_high_glucose or bmi_status == "fat":
            return "middle risk"
        else:
            return "low risk"
```

```python
class Doctor:
    def __init__(self,name, major):
        self.name = name
        self.major = major
        self.patients= []


    def in_hos(self,*p):
        for each in p:
            self.patients.append(each)
        


    def out_hos(self, q):
        self.patients.remove(q) if q in self.patients else print('没有这个患者')

    def basic(self):
        print(f'Name is {self.name}')
        print(f'Major is {self.major}')
        print(f'patients are {[p.name for p in self.patients]}')
        for p in self.patients:
            p.summary()
    
```

```python
Tom = Patient('Tom', 67, 23)
Tom.high_glucose(50)
Tom.summary()
Tom.risk_note()
```

```python
M = Patient('merry', 67, 23)
D = Patient('dove', 67, 23)
```

```python
Bian = Doctor('Bian', 'inner')
Wang = Doctor('Wang', 'outer')
Bian.in_hos(Tom, M, D)
Bian.basic()
Bian.out_hos(M)
Wang.in_hos(M)
Wang.basic()
```
